# 1. Exploratory Data Analysis

Builds the full engineered dataset once (region mapping, derived metrics, normalized columns) via
`international_airlines.features.build_features`, and saves it so notebooks 2 and 3 read the exact
same columns -- this is the fix for the original prototype's broken cross-notebook dependency
(see the top-level README's "What changed" section).

```python
import sys

sys.path.insert(0, '../src')

from pathlib import Path

from international_airlines.eda import (
    plot_region_distribution,
    plot_top_countries,
    plot_yearly_trends,
    top_countries_by_passengers,
)
from international_airlines.features import build_features

RESULTS_DIR = Path('../results')

df = build_features('../data/city_pairs.csv')
df.to_csv(RESULTS_DIR / 'processed_city_pairs.csv', index=False)
print(f"Rows: {len(df)}")
df.head()
```

```python
top_countries_by_passengers(df)
```

```python
plot_top_countries(df, RESULTS_DIR)
plot_region_distribution(df, RESULTS_DIR)
plot_yearly_trends(df, RESULTS_DIR)
print('Saved plots to', RESULTS_DIR)
```
